```python
# Importing Libraries

import joblib
import numpy as np
```

```python
# Loading the Trained Model

model = joblib.load("House-Prediction-model")
```

```python
# Defining a Function for Yes/No Inputs

def yes_no(question):
    while True:
        answer = input(question + "(yes/no):").strip().lower()
        if answer == "yes":
            return 1
        elif answer == "no":
            return 0
        else:
            print("Please enter yes or no")
```

```python
# Getting Numerical Inputs from the User

area = int(input("Enter house area: "))
bedrooms = int(input("Enter number of bedrooms: "))
bathrooms = int(input("Enter number of bathrooms: "))
stories = int(input("Enter number of stories: "))
```

```python
# Getting Categorical Yes/No Inputs

mainroad = yes_no("Is the house near the main road?")
guestroom = yes_no("Does the house have a guestroom?")
basement = yes_no("Does the house have a basement?")
hotwaterheating = yes_no("Does the house have hot water heating?")
airconditioning = yes_no("Does the house have air conditioning?")
```

```python
# Getting Parking Information

parking = int(input("Enter number of parking spaces: "))
```

```python
# Checking Preferred Area

prefarea = yes_no("Is the house in a preferred area?")
```

```python
# Converting Furnishing Status into Numerical Values

while True:

    furnishingstatus = input(
        "Enter furnishing status (unfurnished/semi-furnished/furnished): "
    ).strip().lower()

    furnishing_map = {
        "unfurnished": 0,
        "semi-furnished": 1,
        "furnished": 2
    }

    if furnishingstatus in furnishing_map:
        furnishingstatus = furnishing_map[furnishingstatus]
        break

    else:
        print("Invalid furnishing status. Try again.")
```

```python
# Preparing Input Data for the Model

house_data = np.array([[
    area,
    bedrooms,
    bathrooms,
    stories,
    mainroad,
    guestroom,
    basement,
    hotwaterheating,
    airconditioning,
    parking,
    prefarea,
    furnishingstatus
]])
```

```python
# Predicting the House Price

prediction = model.predict(house_data)
```

```python
# Displaying the Final Prediction

print("\nEstimated House Price:")
print(f"{prediction[0]:,.2f}")
```
